```python
import pandas as pd
import yfinance as yf
from datetime import datetime
import time

def download_with_retry(ticker, start_date, end_date, max_retries=3, delay=2):
    for attempt in range(max_retries):
        try:
            if ticker == 'GOOG':
                ticker = 'GOOGL'
            stock = yf.Ticker(ticker)
            df = stock.history(start=start_date, end=end_date)
            if df.empty:
                time.sleep(delay)
                continue
            return df
        except Exception as e:
            if attempt < max_retries - 1:
                time.sleep(delay)
            else:
                return None

tech_list = ['AAPL', 'GOOG', 'MSFT', 'AMZN']
company_name = ["APPLE", "GOOGLE", "MICROSOFT", "AMAZON"]
end = datetime.now()
start = datetime(end.year - 7, end.month, end.day)
company_list = []

for stock, name in zip(tech_list, company_name):
    df = download_with_retry(stock, start, end)
    if df is not None and not df.empty:
        df['company_name'] = name
        df['ticker'] = stock
        df.to_csv(f"{stock}_data.csv")
        company_list.append(df)

if company_list:
    combined_df = pd.concat(company_list, axis=0)
    combined_df.to_csv("combined_tech_data.csv")
```
